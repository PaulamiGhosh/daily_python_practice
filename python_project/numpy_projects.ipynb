#### *Student Score Checker*

```python
mat = np.random.randint(1,100,(12,5))
print(mat)
```

```python
first_row = mat[:,3]
print(np.mean(mat, axis = 1))  # row wise mean
print(np.mean(mat, axis = 0))  # col wise mean
```

### *complete blood count analyzer*

```python
# hemoglobin range 12 - 15
# rbc count 4m - 6m 
# wbc count 4.5k - 11k
# platelet count 150k - 450k
# hematocrit range 35.5% - 50%
# mcv range 80 - 100
# mch range 27 - 33
# mchc range 32% - 36%
```

```python
sex = np.random.randint(0,2,(10,1))
age = np.random.randint(10,61, (10,1))
heme = np.random.randint(1,25,(10,1))
rbc = np.random.randint(1,10, (10,1))
wbc = np.random.randint(1,20, (10,1))
plat = np.random.randint(10,900, (10,1))
hcrit = np.random.randint(1,100,(10,1))
mcv = np.random.randint(50,130, (10,1))
mch = np.random.randint(10,50,(10,1))
mchc = np.random.randint(30,41,(10,1))
```

```python
data = np.hstack((sex,age,heme,rbc,wbc,plat,hcrit,mcv,mch,mchc))
print(data)
```

```python
# hackerrank concatenate in numpy
import numpy as np
n,m,p = map(int,input().split())

lst1 = []
for i in range(n):
    lst1.append(list(map(int, input().split())))
arr1 = np.array(lst1)

lst2 = []
for i in range(m):
    lst2.append(list(map(int, input().split())))
arr2 = np.array(lst2)
print(np.concatenate((arr1,arr2), axis = 0))
```
