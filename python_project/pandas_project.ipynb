```python
wine = pd.read_csv("/home/paulami/programming/daily_python_practice/python_project/winequality-red.csv")
wine.head()   # show first 5 rows
```

```python
# ---------- level 1 -------------------
wine.shape[0] # for no. of rows
wine.shape[1] # for no. of col
list(wine.columns) #all cols in a list
pd.DataFrame(wine.columns) # all cols in a dataframe
```

```python
# ---------- level 2 -------------------
sub_df = pd.DataFrame(wine[['alcohol','pH','quality']])
sub_df.head(10) # display first 10 rows
sub_df.tail(10) # display last 10 rows
sub_df.iloc[20:30] # 20 included 30 is not
sub_df.loc[20:30] # both 20 and 30 are included
sub_df.loc[5] # a specific row 
alc1 = sub_df.loc[1,['alcohol']]
qua10 = sub_df.loc[10,['quality']]
pH20 = sub_df.loc[20,['pH']]
print(alc1,qua10,pH20)
```

```python
# ---------- level 3 -------------------
sub_df.query('quality >= 7')
(sub_df['quality'] >= 7).sum() 
wine.query('alcohol > 12')[['alcohol','quality']] # high alcohol wines
wine.query('quality >= 7')[['alcohol','quality']] # high quality wines
wine.query('quality <= 4')[['alcohol','quality']] # low quality wines
wine.query('alcohol > 12 & quality >= 7')[['alcohol','quality']] # and operation in query
wine.query('pH <= 3.5 | quality >= 7')[['alcohol','pH','quality']] # or operation
```

```python
# ---------- level 4 -------------------
wine.sort_values(by='alcohol',ascending=False) #sorting respect to alcohol
wine.sort_values(by=['quality','alcohol'], ascending=[False,False])[['alcohol','quality','pH']]
```
